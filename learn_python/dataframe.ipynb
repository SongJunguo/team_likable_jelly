```python
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6],
    'C': [7, 8, 9]
})
print(df)
# 单方括号
series_result = df['A']
print(type(series_result))  # <class 'pandas.core.series.Series'>
print(series_result)
# 0    1
# 1    2  
# 2    3
# Name: A, dtype: int64

# 双方括号
dataframe_result = df[['A']]
print(type(dataframe_result))  # <class 'pandas.core.frame.DataFrame'>
print(dataframe_result)
#    A
# 0  1
# 1  2
# 2  3
```
