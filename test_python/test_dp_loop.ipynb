```python

import pandas as pd, numpy as np
f = '/workspace/aircraft_trajectory/team_likable_jelly/opensky_2024_PRC_dataset/segmented_trajs_doublepass_lim20s_inside_v1/segmented_2022-11-22.parquet'
cols = ['flight_id','timestamp','latitude','longitude','altitude']
seg = pd.read_parquet(f, columns=cols)
seg = seg[seg.flight_id == 2571929180002].reset_index(drop=True)
print(seg.loc[740:760])
# 顺便看一下速度峰值
lat = np.radians(seg.latitude.to_numpy())
lon = np.radians(seg.longitude.to_numpy())
ts = seg.timestamp.values.astype('datetime64[ns]')
mask = (~np.isnan(lat)) & (~np.isnan(lon))
lat, lon, ts = lat[mask], lon[mask], ts[mask]
dist = 2*6371*np.arcsin(np.sqrt(np.clip(
    np.sin((lat[1:]-lat[:-1])/2)**2 + np.cos(lat[:-1])*np.cos(lat[1:])*np.sin((lon[1:]-lon[:-1])/2)**2, 0, 1)))
dt = (ts[1:]-ts[:-1]).astype('timedelta64[s]').astype(float)
vel = dist/(dt/3600)
print('max speed km/h:', np.nanmax(vel))
```
